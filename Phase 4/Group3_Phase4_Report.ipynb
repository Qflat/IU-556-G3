**Project Abstract**

The HCDR project focuses on predicting which clients are likely to default on their home credit loans using the datasets provided in the HCDR Kaggle competition. Earlier phases established baseline machine learning models, explored dataset structure, and developed engineering features to strengthen modeling pipelines performances. Tasks throughout the previous phases included feature engineering and model tuning. Specifically, Phase 3 conducted RFM feature engineering. Phase 4 extends this work by implementing a deep learning approach using a multi-layer perceptron built with PyTorch Lightning. The final feature-engineered dataset was scaled, converted into tensors, and loaded through a Lightning DataModule, enabling efficient GPU-accelerated training and monitoring through the TensorBoard. Multiple models were trained and compared throughout the project. While the deep learning model produced stable learning behavior, it did not outperform the previous best modeling pipeline, ensemble model (LightGBM + CatBoost), with best AUC/ROC score of 0.75.

**Introduction**

Phase 4 concludes the HCDR project. This phase introduces the deep learning component of the HCDR project, building on the feature engineering and baseline models developed in earlier phases. The goal is to produce more accurate predictions than previous phases.

*Data Description/Data Lineage*

Throughout all phases we used the data sets provided by the HCDR Kaggle Competition file. During Phase 4, the following datasets were used:

 - Application Data: Includes the majority of information regarding clients, such as the gender, income, family status, education, number of children, amount of income, and whether the client owns a car or not. The application data has split the main training data from the testing data.
- Bureau Data: This dataset contains data on clients’ credit history.
- Bureau_Balance Data: This dataset breaks down the credit history by months.
- Previous_Application Data: Contains information on clients who have had previous applications for loans from Home Credit.
- POS_CASH_BALANCE Data: Breaks down point of sale and cash loan data by months.
- Credit_Card_Balance Data: Contains data for clients who have had a Home Credit credit card.
- Installments_Payment Data: Includes payment histories for clients who have had a previous Home Credit loan.

For Phase 4, we used merged and feature-engineered dataset produced in previous phases. The data was transformed into scaled tensors through a PyTorch Lightning DataModule to serves as the final input for training and validation of the deep learning model.

**Methods**

This phase extended the prior modeling work by implementing a deep-learning classifier using PyTorch Lightning to predict loan default. The final cleaned and feature-engineered datasets from earlier phases was loaded, scaled, and converted into tensors to support neural-network training. A Lightning data module was used to streamline data loading and batching, ensuring that the training, validation, and test sets followed a consistent pipeline. The model intialized GPU accelerated training, defined loss functions, and prepared the model for inference.

A multi-layer perceptron was implemented as the core model architecture, consisting of an input layer matched to the engineered features, multiple hidden layers with nonlinear activations, and a final sigmoid output unit for binary classification. Training was conducted through Lightning's automated training loop, with TensorBoard used to visualize training and validation performance. Predictions were generated on the competition test set. The model's performance was evaluated trhough public and private leaderboard scores from Kaggle. 14 features were used for feature engineering.


**Modeling Pipelines**

Using the fully merged and preprocessed dataset produced in earlier phases, the workflow began by scaling the numerical features and converting them into tensors suitable for neural-network training. A custom Lightning DataModule handled the training, validation, test splits, batching, and device placement. The neural-network model was built as multi-layer perceptron with linear layers, ReLU activations, dropouts for regularization, and a final sigmoid output for binary classification. The Lightning training loop managed forward passes, backpropagation, optimizer steps, and loss computation while integrating TensorBoard logging.

Several training runs were conducted to evaluate how the network learned from the HCDR data. Hyperparameters, such as batch size, learning rate, number of epochs, and dropout percentage, were adjusted across experiments to observe the impact on the models' performance. TensorBoard was used to montior learning curves for traning and validation loss. After the final model completed training, predictions were generated on the competition's test set and submitted to Kaggle.

**Results and Discussion of Results**

For Phase 4, the best validation AUC was 0.74486. The best performing model overall was ensemble (LightGBM + CatBoost). It produced the hgihest ROC-AUC score overall at 0.7526. The best features were: CODE_GENDER_M, EXT_SOURCE_3, NAME_EDUCATION_TYPE_Higher EDUCATION, EXT_SOURCE_2, CODE_GENDER_F, NAME_INCOME_TYPE_Pensioner,NAME_CONTRACT_TYPE_Cash loans, FLAG_OWN_CAR_N,NAME_INCOME_TYPE_Working, NAME_EDUCATION_TYPE_Secondary / secondary special. The Deep Learning Model only achieved a 0.74486 (no hyperparameters). This shows that engineered features where most likely a primary driver in both models for performance.

**Gap Analysis**

Our best-performing pipeline shows a noticeable but reasonable gap when compared with the strongest submissions on the CaDoD and HCDR scoreboards. While our ensemble model achieved a ROC AUC of roughly 0.75, top leaderboard entries typically push several points higher by incorporating far more extensive feature engineering and deeper use of historical application behavior. These high-performing teams often integrate multiple auxiliary datasets, create hundreds of aggregated behavioral indicators, and apply sophisticated tuning approaches such as large-scale randomized searches, Bayesian optimization, or multi-layer model stacking. In contrast, our Phase 4 workflow focused on a streamlined set of engineered variables and moderate optimization, which explains much of the performance difference. Nonetheless, our results indicate that the current pipeline is fundamentally sound and well-calibrated, offering a strong platform that could close the remaining gap through richer feature development, broader cross-table aggregations, and more advanced search over model configurations.
